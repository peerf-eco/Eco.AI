```python
import os, glob, csv
from PIL import Image
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_recall_curve, auc, average_precision_score
from sklearn.preprocessing import label_binarize
from IPython.display import display, Markdown
import warnings
warnings.filterwarnings("ignore")

ORL_PATH    = r"orl_faces"
FRUITS_PATH = r"fruits-360_100x100/fruits-360/Training"
RTSD_PATH   = r"gtrsd/Train"

TARGET_SIZE = (100,100)
GRAY = True
TEST_SIZE = 0.2
RANDOM_STATE = 42
TRAIN_SIZE_FRACS = [0.1, 0.25, 0.5, 0.75, 1.0]
PCA_N_COMPONENTS = 100
KNN_NEIGHBORS = 3

MAX_PER_CLASS_FRUITS = 30
MAX_PER_CLASS_ORL = None
MAX_PER_CLASS_RTSD = 40


def _get_pil_resample():
    if hasattr(Image, "Resampling"):
        return Image.Resampling.LANCZOS
    if hasattr(Image, "LANCZOS"):
        return Image.LANCZOS
    if hasattr(Image, "ANTIALIAS"):
        return Image.ANTIALIAS
    return Image.BICUBIC

_RESAMPLE = _get_pil_resample()

def load_images_flexible(root_folder, target_size=(100,100), gray=True, max_per_class=None):
    if not os.path.exists(root_folder):
        raise FileNotFoundError(f"Path '{root_folder}' does not exist.")
    subdirs = sorted([d for d in os.listdir(root_folder) if os.path.isdir(os.path.join(root_folder, d))])
    X = []; y = []; files = []
    if subdirs:
        for cls in subdirs:
            cls_folder = os.path.join(root_folder, cls)
            patterns = ["*.jpg","*.png","*.jpeg","*.pgm","*.bmp"]
            imgs = []
            for p in patterns:
                imgs.extend(glob.glob(os.path.join(cls_folder, p)))
            imgs = sorted(imgs)
            if not imgs:
                continue
            if max_per_class is not None:
                imgs = imgs[:max_per_class]
            for img_path in imgs:
                try:
                    img = Image.open(img_path); img.load()
                    img = img.convert("L") if gray else img.convert("RGB")
                    img = img.resize(target_size, resample=_RESAMPLE)
                    arr = np.asarray(img, dtype=np.uint8)
                    X.append(arr.flatten()); y.append(cls); files.append(img_path)
                except Exception as e:
                    print(f"Warning: failed to load {img_path}: {e}")
        if len(X) == 0:
            return np.empty((0,)), np.empty((0,)), []
        return np.vstack(X).astype(np.float32), np.array(y), files

   
def compute_multiclass_map(y_true, scores, classes):
    if len(y_true) == 0 or scores.size == 0:
        return {c: np.nan for c in classes}, np.nan
    lb = label_binarize(y_true, classes=classes)
    aps = {}
    for i, cls in enumerate(classes):
        y_bin = lb[:, i]; s = scores[:, i]
        if np.all(y_bin == 0):
            aps[cls] = np.nan
        else:
            aps[cls] = average_precision_score(y_bin, s)
    ap_list = [v for v in aps.values() if not np.isnan(v)]
    mAP = np.mean(ap_list) if ap_list else np.nan
    return aps, mAP


def run_one_dataset_nb(name, path, max_per_class=None):
    display(Markdown(f"## Dataset: **{name}**"))
    X, y, files = load_images_flexible(path, target_size=TARGET_SIZE, gray=GRAY, max_per_class=max_per_class)
    if X.size == 0:
        display(Markdown("**No images loaded or no labels found.**"))
        return None
    display(Markdown(f"- Samples loaded: **{len(y)}**; classes: **{len(np.unique(y))}**"))
    display(Markdown("**Примеры файлов:**"))
    display(pd.Series(files[:6]))

    X = X / 255.0
    X_train_all, X_test, y_train_all, y_test = train_test_split(X, y, test_size=TEST_SIZE, stratify=y if len(np.unique(y))>1 else None, random_state=RANDOM_STATE)
    classes = sorted(np.unique(y))
    display(Markdown(f"- Train pool: **{len(y_train_all)}**, Test: **{len(y_test)}**"))

    results = []
    last_scores_mat = None
    confusion_for_full = None; aps_for_full = None; pca_visual_proj = None

    for frac in TRAIN_SIZE_FRACS:
        n_train = int(len(y_train_all) * frac)
        if n_train < len(classes):
            display(Markdown(f"- skipping frac **{frac:.2f}** (n_train={n_train}) — not enough samples"))
            continue
        rng = np.random.RandomState(RANDOM_STATE); idx_all = np.arange(len(y_train_all)); rng.shuffle(idx_all)
        sel_idx = idx_all[:n_train]
        X_train = X_train_all[sel_idx]; y_train = y_train_all[sel_idx]

        n_comp = min(PCA_N_COMPONENTS, max(1, n_train-1), X_train.shape[1])
        pca = PCA(n_components=n_comp, svd_solver='randomized', random_state=RANDOM_STATE)
        pca.fit(X_train)
        X_train_p = pca.transform(X_train); X_test_p = pca.transform(X_test)

        clf = KNeighborsClassifier(n_neighbors=KNN_NEIGHBORS); clf.fit(X_train_p, y_train)
        y_pred = clf.predict(X_test_p)
        try:
            probs = clf.predict_proba(X_test_p); clf_classes = list(clf.classes_)
            scores_mat = np.zeros((len(y_test), len(classes)))
            for j, cls in enumerate(classes):
                if cls in clf_classes:
                    idx = clf_classes.index(cls); scores_mat[:, j] = probs[:, idx]
                else:
                    scores_mat[:, j] = 0.0
        except Exception:
            scores_mat = np.zeros((len(y_test), len(classes)))
        last_scores_mat = scores_mat

        acc = accuracy_score(y_test, y_pred)
        cm = confusion_matrix(y_test, y_pred, labels=classes)
        aps, mAP = compute_multiclass_map(y_test, scores_mat, classes)

        results.append({"train_frac": frac, "n_train": n_train, "n_test": len(y_test), "n_classes": len(classes), "pca_n_components": n_comp, "accuracy": acc, "mAP": mAP})
        display(Markdown(f"- frac **{frac:.2f}** | n_train={n_train} | acc={acc:.4f} | mAP={mAP:.4f}"))

        if abs(frac-1.0) < 1e-9:
            confusion_for_full = (cm, classes); aps_for_full = aps
            try:
                pca2 = PCA(n_components=2, random_state=RANDOM_STATE).fit(np.vstack([X_train_all, X_test]))
                proj = pca2.transform(np.vstack([X_train_all, X_test])); y_all = np.concatenate([y_train_all, y_test]); pca_visual_proj = (proj, y_all)
            except Exception:
                pca_visual_proj = None

    df_res = pd.DataFrame(results).sort_values("train_frac").reset_index(drop=True)
    display(Markdown("### Summary table")); display(df_res)

    # learning curve
    if not df_res.empty:
        plt.figure(figsize=(6,3))
        plt.plot(df_res['train_frac'], df_res['accuracy'], marker='o', label='accuracy')
        plt.plot(df_res['train_frac'], df_res['mAP'], marker='s', label='mAP')
        plt.xlabel('train fraction'); plt.ylabel('metric'); plt.title(f"Learning curve ({name})"); plt.grid(True); plt.legend(); plt.show()

    # confusion matrix
    if confusion_for_full is not None:
        cm, classes = confusion_for_full
        display(Markdown("### Confusion matrix (full train)"))
        plt.figure(figsize=(6,5)); plt.imshow(cm, interpolation='nearest', aspect='auto'); plt.colorbar()
        plt.xlabel('pred'); plt.ylabel('true'); plt.xticks(ticks=np.arange(len(classes)), labels=classes, rotation=90); plt.yticks(ticks=np.arange(len(classes)), labels=classes)
        thresh = cm.max()/2. if cm.size else 0
        for i in range(cm.shape[0]):
            for j in range(cm.shape[1]):
                plt.text(j,i,format(int(cm[i,j]),'d'),ha='center', color='white' if cm[i,j]>thresh else 'black')
        plt.tight_layout(); plt.show()

    # per-class AP
    if aps_for_full is not None:
        display(Markdown("### Per-class AP (full train)"))
        display(pd.Series(aps_for_full).sort_values(ascending=False).to_frame("AP"))

    # PR-curves for up to 3 classes
    if last_scores_mat is not None and last_scores_mat.size>0:
        display(Markdown("### PR-curves (up to 3 classes)"))
        classes_to_plot = classes[:min(3,len(classes))]
        plt.figure(figsize=(5,4))
        for cls in classes_to_plot:
            y_bin = (y_test==cls).astype(int); s = last_scores_mat[:, classes.index(cls)]
            if np.sum(y_bin)==0: continue
            prec, rec, _ = precision_recall_curve(y_bin, s); ap = auc(rec,prec)
            plt.plot(rec,prec,label=f"{cls} (AP={ap:.3f})")
        plt.xlabel('Recall'); plt.ylabel('Precision'); plt.legend(fontsize='small'); plt.grid(True); plt.show()

    # PCA scatter
    if pca_visual_proj is not None:
        proj, y_all = pca_visual_proj
        display(Markdown("### PCA(2) projection (train+test)"))
        plt.figure(figsize=(6,4))
        uniq = sorted(np.unique(y_all))
        for c in uniq:
            mask = (y_all==c)
            plt.scatter(proj[mask,0], proj[mask,1], s=6, alpha=0.6, label=str(c))
        plt.legend(bbox_to_anchor=(1.02,1), markerscale=3, fontsize='small')
        plt.title("PCA(2) projection"); plt.tight_layout(); plt.show()

    return {"results_df": df_res, "confusion_full": confusion_for_full, "per_class_ap": aps_for_full, "pca_proj": pca_visual_proj}

results = {}
results['orl'] = run_one_dataset_nb("ORL_faces", ORL_PATH, max_per_class=MAX_PER_CLASS_ORL)
results['fruits'] = run_one_dataset_nb("Fruits360_100x100", FRUITS_PATH, max_per_class=MAX_PER_CLASS_FRUITS)
results['rtsd'] = run_one_dataset_nb("RTSD", RTSD_PATH, max_per_class=MAX_PER_CLASS_RTSD)
```
